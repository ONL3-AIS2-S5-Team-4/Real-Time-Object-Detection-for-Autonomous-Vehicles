## 1. Setup & Configuration
Run this first to define paths and import libraries.

```python
from ultralytics import YOLO
import cv2
import os

# --- Paths & Config ---
MODEL_PATH = r"best135K.onnx"
IMAGE_INPUT = r"12.jpg"
VIDEO_INPUT = r"Generate_Night_Rainy_Traffic_Video.mp4"
OUTPUT_DIR = r"runs/detect_new"

os.makedirs(OUTPUT_DIR, exist_ok=True)
CONF_THRESH = 0.25
IMG_SIZE = 416
```

## 2. Load Model
Loads the model into memory.

```python
print(f"Loading model: {MODEL_PATH}...")
model = YOLO(MODEL_PATH)
print("Model loaded successfully.")
```

## 3. Image Detection
Runs inference on the single image, saves it, and opens a window.

```python
print(f"Processing image: {IMAGE_INPUT}")

# Inference
results = model.predict(IMAGE_INPUT, imgsz=IMG_SIZE, conf=CONF_THRESH)
img_out = results[0].plot()

# Save
save_path = os.path.join(OUTPUT_DIR, "output_image.jpg")
cv2.imwrite(save_path, img_out)
print(f"Saved at: {save_path}")

# Display
cv2.imshow("Image Detection", img_out)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 4. Video Detection
Runs inference on the video frame-by-frame. Press 'q' to stop early.

```python
print(f"Processing video: {VIDEO_INPUT}")

cap = cv2.VideoCapture(VIDEO_INPUT)

# Setup Video Writer
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
save_path = os.path.join(OUTPUT_DIR, "output_video.mp4")
out = cv2.VideoWriter(save_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, (width, height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret: break

    # Inference
    results = model.predict(frame, imgsz=IMG_SIZE, conf=CONF_THRESH, verbose=False)
    frame_out = results[0].plot()

    # Save & Display
    out.write(frame_out)
    cv2.imshow("Video Detection", frame_out)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
print(f"Video saved at: {save_path}")
```
